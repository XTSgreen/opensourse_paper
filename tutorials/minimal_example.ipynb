# Minimal UOT-IOT example

This notebook demonstrates the public model contract on a small synthetic transition panel. It uses no test target information during fitting beyond the retrospective contract.

```python
import numpy as np
from iot_benchmark.models.uot_iot import UOTIOTBenchmarkModel

source = np.array([[8, 2], [2, 8], [7, 3], [3, 7]], dtype=float)
target = np.array([[6, 4], [3, 7], [5, 5], [4, 6]], dtype=float)
groups = np.array(['replicate_a', 'replicate_a', 'replicate_b', 'replicate_b'])
cost = np.array([[0.0, 1.0], [1.0, 0.0]])
model = UOTIOTBenchmarkModel().fit(source, target, groups, seed=20260916)
model.direction()
```

```python
prediction = model.predict_state(source)
prediction.shape, prediction.sum(axis=1)
```
